## **Comparação entre as tabelas de Multas e Acidentes**

Apesar de possuírem objetivos diferentes, as tabelas **Multas** e **Acidentes** apresentam algumas colunas com informações semelhantes.

### Informações em comum

- Ambas possuem uma coluna de **data**, no formato `dd/mm/aaaa`.
- Possuem informações para identificar o **local da ocorrência**, como:
  - **BR**
  - **KM da BR**
- Ambas informam o **município** onde ocorreu o acidente ou a infração.
- Também possuem a **Unidade Federativa (UF)** correspondente à ocorrência.

### Informações específicas

Embora existam semelhanças entre algumas variáveis, cada tabela possui colunas específicas relacionadas ao seu objetivo.

Na tabela de **Acidentes**, são utilizadas colunas como:

- `causa_acidente`
- `tipo_acidente`
- Quantidade de feridos, incluindo:
  - Feridos leves
  - Feridos graves
  - Feridos ignorados
  - Mortos
- Tipo da pista
- Condição meteorológica
- Outras informações relacionadas às características e consequências do acidente.

Já a tabela de **Multas** possui informações relacionadas à fiscalização e às infrações de trânsito, como:

- Informações sobre a abordagem do veículo
- Sentido do tráfego
- Data de início e fim da vigência
- Medição realizada
- `descricao_abreviada`, utilizada para identificar a infração cometida
- Outras informações específicas relacionadas à aplicação das multas.

Embora as duas tabelas compartilhem informações relacionadas ao **tempo e ao local das ocorrências**, suas demais variáveis são voltadas para objetivos distintos: a tabela **Acidentes** apresenta informações sobre ocorrências e suas consequências, enquanto a tabela **Multas** está relacionada à fiscalização e às infrações de trânsito.

## **Classificação de variáveis categóricas**

`dia_semana`

A variável `dia_semana` é **qualitativa nominal**, pois representa categorias correspondentes aos dias da semana, como segunda-feira, terça-feira e quarta-feira.

Embora os dias possam ser organizados em uma sequência cronológica, essa sequência não representa uma hierarquia ou uma ordem de grandeza entre as categorias. Não é possível afirmar que segunda-feira é maior ou menor que terça-feira.

`fase_dia`

A variável `fase_dia` é **qualitativa ordinal**, pois representa categorias relacionadas ao momento do dia, como `Pleno dia`, `Anoitecer` e `Plena noite`.

Essas categorias podem ser organizadas em uma sequência relacionada a luminosidade e a progressão do horário do dia. Entretanto, não existe uma distância numérica definida entre elas.

Essa variável pode ser armazenada como texto ou categoria, mas pode receber uma ordem específica. Por exemplo, no Pandas, seria possível definir uma ordem entre as categorias para facilitar análises e gráficos.

`idade`

A variável `idade` é **quantitativa contínua**, pois representa uma medida numérica que pode, em sua natureza, assumir qualquer valor dentro de um intervalo.

Por exemplo, uma pessoa pode ter 17 anos, 17,5 anos ou 17 anos e 3 meses. Mesmo quando a idade é registrada no dataset apenas em anos completos, a variável continua sendo naturalmente contínua.

`veiculos`

A variável `veiculos` é **quantitativa discreta**, pois representa uma contagem da quantidade de veículos envolvidos em uma ocorrência.

Ela pode assumir valores inteiros, como 1, 2, 3 ou 4 veículos, mas não valores intermediários. Por exemplo, não é possível existir 2,5 veículos envolvidos em um acidente.

## **Codificação das varáveis categóricas**

Para substituir os valores categóricos de `fase_dia`, eu optei por usar uma função e através dessa função usar a função de mapeamento `apply`. Respeitando a gradação da luz do dia.

```python
def substituicao(x):
    if x== 'Amanhecer':
        return 1
    elif x== 'Pleno dia':
        return 2
    elif x== 'Anoitecer':
        return 3
    elif x== 'Plena noite':
        return 4
    elif not x:
        return None

tabela_acidentes['fase_dia']= tabela_acidentes['fase_dia'].apply(substituicao)
```

## **Categorização da coluna `idade`**

Apliquei o mesmo principio da tarefa anterior, usando funções e mapeamentos.

```python
def substituicao(x):
    if 0<=x<18:
        return 'Menores de Idade'
    elif 18<=x<30:
        return 'Jovens'
    elif 30<=x<60:
        return 'Adultos'
    elif 60<=x<120:
        return 'Idosos'

tabela_acidentes['faixa_etaria']= tabela_acidentes['idade'].apply(substituicao)
```

## **Frequencia das variáveis categóricas**

```python
fi= tabela_acidentes['faixa_etaria'].value_counts()
fr= (tabela_acidentes['faixa_etaria'].value_counts(normalize=True)*100)
Fi= fi.cumsum()
Fr= fr.cumsum()
total= tabela_acidentes['faixa_etaria'].value_counts().sum()
```

```python
tabela= pd.DataFrame(
    {'fi': fi, 'fr(%)': fr, 'Fi': Fi, 'Fr(%)': Fr},
    index=['Adultos', 'Menores de Idade', 'Jovens', 'Idosos']
)

total= pd.DataFrame({
    'fi': fi.sum(),
    'fr(%)': fr.sum(),
    'Fi': '-',
    'Fr(%)': '-' 
}, index=['Total'])

pd.concat([tabela, total])
```
